## Imports

The core of sir3stoolkit is a Python wrapper around basic functionality of SIR 3S, offering a low-level access to the creation, modification and simulation of SIR 3S models. In the future pure python subpackages may be added.

```python
from sir3stoolkit.core import wrapper
```

```python
sir3stoolkit
```

The Toolkit has to be initialized with reference to a SIR 3S (SirGraf) installation using [Initialize_Toolkit()](https://3sconsult.github.io/sir3stoolkit/modules.html#sir3stoolkit.core.wrapper.Initialize_Toolkit).

```python
wrapper.Initialize_Toolkit(SIR3S_SIRGRAF_DIR)
```

## Sir3S_Toolkit.dll Version

Some sir3stoolkit versions require replacing the Sir3S_Toolkit.dll in the SirGraf directory with the one shipped in the package (see [Releases](https://3sconsult.github.io/sir3stoolkit/releases.html)). From sir3stoolkit 90.15.26 on, the DLL carries a file version, so `Initialize_Toolkit()` checks this automatically: it logs the file version of the DLL in the SirGraf directory (see above) and warns if it differs from the DLL shipped with the installed sir3stoolkit version. The file version can also be read with [Get_Toolkit_Dll_File_Version()](https://3sconsult.github.io/sir3stoolkit/references/sir3stoolkit.core.html#sir3stoolkit.core.wrapper.Get_Toolkit_Dll_File_Version) - by default from the SirGraf directory, or from any Sir3S_Toolkit.dll path. DLLs before 90.15.26 report 1.0.0.0.

```python
wrapper.Get_Toolkit_Dll_File_Version()
```

## Persisting the SirGraf Path

Hardcoding `SIR3S_SIRGRAF_DIR` in every script is repetitive. Instead, the path can be written once to the package's config file via [Write_SirGraf_Config_Path()](https://3sconsult.github.io/sir3stoolkit/references/sir3stoolkit.core.html#sir3stoolkit.core.wrapper.Write_SirGraf_Config_Path) - after that, [Initialize_Toolkit()](https://3sconsult.github.io/sir3stoolkit/modules.html#sir3stoolkit.core.wrapper.Initialize_Toolkit) can be called with no argument at all and will resolve the path from that config file automatically (see its docstring for the full resolution order). [Read_SirGraf_Config_Path()](https://3sconsult.github.io/sir3stoolkit/references/sir3stoolkit.core.html#sir3stoolkit.core.wrapper.Read_SirGraf_Config_Path) reads back what is currently configured, or `None` if nothing is set yet.

```python
wrapper.Write_SirGraf_Config_Path(SIR3S_SIRGRAF_DIR)
```

```python
wrapper.Read_SirGraf_Config_Path()
```

With the path persisted, `Initialize_Toolkit()` no longer needs `SIR3S_SIRGRAF_DIR` explicitly:

```python
wrapper.Initialize_Toolkit()  # basePath omitted - resolved from the config file written above
```
