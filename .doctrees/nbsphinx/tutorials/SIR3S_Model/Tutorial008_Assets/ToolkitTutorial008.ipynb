## Set_Raise_On_Error()

When the SIR 3S Toolkit is used from a standalone Python process (like this notebook), errors reported by SIR 3S raise a [SIR3S_ToolkitError](https://3sconsult.github.io/sir3stoolkit/references/sir3stoolkit.core.html#sir3stoolkit.core.wrapper.SIR3S_ToolkitError). Inside SIR Graf they are only logged and the function returns a value indicating the failure (e.g. False, '-1' or ('', '')). Let's try to set the log file path to a directory that does not exist.

```python
try:
    s3s.SetLogFilePath("C:/does/not/exist/Tutorial8.log")
except wrapper.SIR3S_ToolkitError as e:
    print(f"Caught: {e}")
```

We can use the function [Set_Raise_On_Error()](https://3sconsult.github.io/sir3stoolkit/references/sir3stoolkit.core.html#sir3stoolkit.core.wrapper.Set_Raise_On_Error) to change this behavior: False only logs errors (the behavior inside SIR Graf), True always raises, None restores the default.

```python
wrapper.Set_Raise_On_Error(False)
s3s.SetLogFilePath("C:/does/not/exist/Tutorial8.log")
```

```python
wrapper.Set_Raise_On_Error(None)
```

Errors of core calls made inside mantle methods are always only logged, as the mantle handles them itself.

Note that not every invalid input is reported as an error by SIR 3S. Check the stored values with GetValue() (Tutorial 2) and run ExecuteModelValidation() (Tutorial 9) before relying on a model you built or changed.

## EnableOrDisableOutputComments()

We can use the function [EnableOrDisableOutputComments()](https://3sconsult.github.io/sir3stoolkit/references/sir3stoolkit.core.html#sir3stoolkit.core.wrapper.SIR3S_Model.EnableOrDisableOutputComments) to turn output comments on or off. Default is True.

```python
s3s.EnableOrDisableOutputComments(outputComments=True)
```

```python
print(s3s.InsertElement(s3s.ObjectTypes.Node, "-1"))
```

```python
s3s.EnableOrDisableOutputComments(outputComments=False)
```

```python
print(s3s.InsertElement(s3s.ObjectTypes.Node, "-1"))
```

As you can see, after the output comments were disabled the function does not print an output. The printed tk is the return value of the function that is displayed in Jupyter Notebooks.

```python
s3s.EnableOrDisableOutputComments(outputComments=True)
```

## AllowSirMessageBox()

We can use the function [AllowSirMessageBox()](https://3sconsult.github.io/sir3stoolkit/references/sir3stoolkit.core.html#sir3stoolkit.core.wrapper.SIR3S_Model.AllowSirMessageBox) to allow or diallow SIR DB Messages to pop up. Default is True. Generally, not recommeded to turn off, as it could mean missing out on deep error messages.

```python
s3s.AllowSirMessageBox(bAllow=True)
```

```python
dbFilePath2=r"Toolkit_Tutorial8_PotsdamVersionModel.db3"
```

We will try to open a model created in Potsdam Version of SIR 3S. A message window should pop up that asks whether you want the model to be migrated to your current SIR 3S version. Please choose to not migrate. The cell will fail to execute.

```python
s3s.OpenModel(dbName=dbFilePath2,
              providerType=s3s.ProviderTypes.SQLite,
              Mid="M-1-0-1",
              saveCurrentlyOpenModel=False,
              namedInstance="",
              userID="",
              password="")
```

Now we will attempt the same but with message boxes turned off.

```python
s3s.AllowSirMessageBox(bAllow=False)
```

```python
s3s.OpenModel(dbName=dbFilePath2,
              providerType=s3s.ProviderTypes.SQLite,
              Mid="M-1-0-1",
              saveCurrentlyOpenModel=False,
              namedInstance="",
              userID="",
              password="")
```

As you can see in your file explorer this time the model was migrated without asking. In general during model creation/opening the message boxes should be left on. 

This method also works if you want to open a model just to view it, while it is already open in SIR Graf.
